```python
%%writefile hyp.yaml
# Hyperparameters for YOLOv8 training-It controls how the model learns and how the input images are augmented
lr0: 0.01 #	Initial learning rate (0.01)-Controls how fast weights update at the start
lrf: 0.01 #Final learning rate fraction (0.01)-Defines how much the LR decays by the end of training
momentum: 0.937 #Momentum helps training move smoothly toward the goal instead of zig-zagging and wasting steps.
weight_decay: 0.0005 #Prevents overfitting by penalizing large weights
warmup_epochs: 3.0 #Gradually increases LR for first few epochs to avoid unstable training at start
warmup_momentum: 0.8 #Starts with lower momentum to allow faster adaptation early on.
warmup_bias_lr: 0.1 #Allows bias terms to learn faster at the start.

hsv_h: 0.015 #Slightly changes color hue to make model robust to lighting/camera differences.
hsv_s: 0.7 #Adjusts intensity of colors to handle different brightness levels
hsv_v: 0.4 #Changes image brightness for robustness in varying lighting.
degrees: 5.0 #Helps detect objects at slightly rotated angles.
translate: 0.1 #Trains model to detect objects even if they appear off-center.
scale: 0.5 #Makes model handle objects at different distances/sizes.
shear: 0.0 #Distorts image to simulate different viewing angles.
perspective: 0.0 #Tilts image as if viewed from different positions.
flipud: 0.0 #Flips image upside down (disabled here since rare in natural datasets).
fliplr: 0.5 #50% chance to mirror image horizontally
mosaic: 1.0 #Combines 4 images into 1 for better small-object detection and varied contexts.
mixup: 0.2 #Blends two images together to improve robustness and prevent overfitting.
```

```python
from ultralytics import YOLO

# Load pretrained YOLOv8m
model = YOLO('yolov8m.pt')

# Train with custom hyp.yaml
model.train(
    data='/content/dataset/data.yaml',
    epochs=50,
    imgsz=960,
    cfg='/content/hyp.yaml'  # <-- Use cfg instead of hyp
)

```
